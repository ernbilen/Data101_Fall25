## Jupyter Notebook aka `.ipynb` documents

# For loops
For loops are very useful for repetitive tasks.

```python
for i in range(0,5):
    print('This gets repeated')
```

```python
for i in range(0,10):
    print(i*'*')
```

```python
# fancier, more beautiful
for i in range(1, 10):
    spaces = " " * (10 - i)     # leading spaces shrink each row
    stars  = "*" * (2*i - 1)    # 1, 3, 5, ..., 21 stars
    print(spaces + stars)
```

```python
# numpy is python's numeric computing library
# import as below while giving it a nickname: np (work smarter not harder)
import numpy as np 
# if you don't have it installed use ! pip install numpy
# you only have to install a library once

# Simulate 100 student exam scores with an average of 75 and SD of 10,
# then show the first 6 scores
np.random.seed(0) # make results reproducible
scores = np.random.normal(75, 10, 100)  # mean=75, sd=10, n=100
```

```python
print(scores) # print function to inspect your object
```

```python
type(scores) # type() function shows the object is a numpy array aka a vector
```

```python
# alternatively, something like
type([1,2,3])
```

```python
# there is also dictionaries
type({'firstitem':'book','seconditem':'pencil'})
```

```python
# round() function to the scores to 1 decimal place
rounded_scores = np.round(scores, 1)
rounded_scores
```

```python
# How do you select a particular element in a vector: use []
rounded_scores[0] # indexing in python starts from 0 [insert meme]
```

```python
# you can get a subset of numbers:
rounded_scores[0:5]
```

```python
# you can do math like
(rounded_scores[0] + rounded_scores[1])/2
```

```python
# or work smarter not harder
np.mean([rounded_scores[0],rounded_scores[1]]) # np.mean() function from the numpy library, note the second set of []
# becase the np.mean() function needs an array and eg 2,3 is not array but [2,3] is
```

## Importing data with pandas

You can load CSV data straight from a URL. This is similar to `readr::read_csv` in R. Below is a TidyTuesday example URL, loaded with `pandas.read_csv`. You do not need to download the file manually if it is public on GitHub raw links.

```python
import pandas as pd
pd.set_option('display.max_columns', None) # To display all columns

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2022/2022-01-25/details.csv"
board_games = pd.read_csv(url)
board_games.head()
```

## Visualizations with matplotlib

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(5,3))
plt.hist(popular['owned'],bins=250,color='lightseagreen',edgecolor='black')
plt.xlim(0,20000)
plt.xlabel('# of owners') # [insert meme]
plt.ylabel('count')
plt.savefig("ownerscount.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# cleaning categories
popular['category_cleaned'] = popular['boardgamecategory'].str.split(",").str[0].str.strip("[]' ") # you can ignore this
popular['category_cleaned']
```

```python
counts = popular['category_cleaned'].value_counts() # super useful function
counts = counts[counts.values>50] # at least 100 board games in a category
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
plt.bar(counts.index,counts.values) # index has the categories, values has the counts
plt.xlabel('category') # [insert meme]
plt.ylabel('count')
plt.xticks(rotation=90)
plt.show()
```

```python
board_games.yearpublished.describe()
```

```python
# do people hold old or new games more?
# how to find the average number of owners by year? Use groupby(), super important function
grouped = board_games.groupby('yearpublished').mean()['id'].reset_index().rename(columns={'id':'owners'})
grouped = grouped[grouped.yearpublished>1900]
grouped
```

```python
# make y-axis look nicer?
grouped['ownersinthousand'] = grouped.owners/1000

plt.plot(grouped.yearpublished,grouped.ownersinthousand)
plt.xticks(np.arange(1900,2024,10),rotation=45)
plt.xlabel('year')
plt.ylabel('# of onwers, in thousands')
plt.show()
```
